```python
import pandas as pd


# This notebook creates a table of excluded papers from the systematic literature review.
# The table is saved as a tex file and can be included in the paper as table in the appendix.
```

```python
df = pd.read_csv('review_523015_excluded_csv_20250714191040.csv')
df
```

```python
# Select and clean columns
table_df = df[["Title", "Authors", "Published Year", "Notes"]]
table_df["Notes"] = (
    table_df["Notes"]
    .str.replace("Exclusion reason: ", "", regex=False)
    .str.replace("; ", "", regex=False)
)

# Make title column title case
table_df["Title"] = table_df["Title"].str.title()

table_df.columns = ["Title", "Authors", "Year", "Reason for Exclusion"]

# Add index column starting at 1
table_df.insert(0, "No.", range(1, len(table_df) + 1))

# Convert to LaTeX with formatting
latex_table = table_df.to_latex(
    index=False,
    escape=True,
    longtable=True,
    column_format="r p{5cm} p{4cm} p{1cm} p{5cm}",
    caption="List of Studies Excluded in Full-text Review and Reasons"
)

# Write to file with smaller font for fitting the page
with open("excluded_studies_table.tex", "w") as f:
    f.write("\\begin{small}\n")
    f.write(latex_table)
    f.write("\\end{small}\n")
```
