```python
import csv
import pandas as pd
```

```python
# Replace with your actual .txt file path
file_path = "250602_slr_data_extraction.csv"

# Read using proper quoting to avoid splitting commas inside quoted fields
with open(file_path, newline='', encoding='utf-8') as f:
    reader = csv.reader(f, delimiter=",", quotechar='"')
    rows = list(reader)

# Use the first row as header
header = rows[0]
data = rows[1:]

# Optional: deduplicate columns if needed (e.g., "Study ID" appears twice)
# Clean header if necessary
clean_header = []
seen = set()
for col in header:
    if col in seen:
        clean_header.append(f"{col} (dup)")
    else:
        clean_header.append(col)
        seen.add(col)

# Create DataFrame
df = pd.DataFrame(data, columns=clean_header)

# Display sample
print(df.head())
```

```python
df
```

```python
# get only rows where Reviewer Name is Consensus
consensus_df = df[df['Reviewer Name'] == 'Consensus']
# Display the shape of the consensus DataFrame
print(consensus_df.shape)
```

```python
# add 'Year'column
consensus_df['Year'] = consensus_df['Study ID'].str.extract(r'(\d{4})').astype(int)

# remove columns Study ID and Title(dup)
consensus_df = consensus_df.drop(columns=['Study ID', 'Title (dup)'])

# rename Study ID(dup) to Study ID
consensus_df = consensus_df.rename(columns={'Study ID (dup)': 'Study ID'})

# Display the cleaned consensus DataFrame
consensus_df.head()
```

```python
# Show value counts for 'What other datasets are used in the study?'
consensus_df['What other datasets are used in the study?'].value_counts()

# Check for if string contains 'None' and ';', if yes then remove 'None' and ';' from the string
consensus_df['What other datasets are used in the study?'] = consensus_df['What other datasets are used in the study?'].apply(
    lambda x: x.replace('; None', '') if "; None" in x else x)

# Check if string contains 'Trend'and remove it
consensus_df['What other datasets are used in the study?'] = consensus_df['What other datasets are used in the study?'].apply(
    lambda x: x.replace('Trend ', '') if 'Trend' in x else x)


# Show value counts for 'What other datasets are used in the study?'
consensus_df['What other datasets are used in the study?'].value_counts()
```

```python
# standardize versions of 'n/a' to 'N/A' in consensus_df for all columns
consensus_df = consensus_df.replace({'Other: n/a': 'Other: N/A', 'Other: na': 'Other: N/A', 'Other: n.a.': 'Other: N/A'}, regex=True)
# Show value counts for 'What other datasets are used in the study?'
consensus_df['What other datasets are used in the study?'].value_counts()
```

```python
# make lower case for 'What is the granularity of the price dataset?'
consensus_df['What is the granularity of the price dataset?'] = consensus_df['What is the granularity of the price dataset?'].str.lower()

# Remove 'Other: ' prefix from granularity_list
consensus_df['What is the granularity of the price dataset?'] = consensus_df['What is the granularity of the price dataset?'].replace({'other: ':''}, regex=True)

# standardize version of "minute"
consensus_df['What is the granularity of the price dataset?'] = consensus_df['What is the granularity of the price dataset?'].replace({'min': 'minute', 'minutes': 'minute'})

# standardize version of 'ticker' to "tick level"
consensus_df['What is the granularity of the price dataset?'] = consensus_df['What is the granularity of the price dataset?'].apply(lambda x: 'tick level' if 'tick' in x.lower() else x)

# replace 'daily' with 'day'
consensus_df['What is the granularity of the price dataset?'] = consensus_df['What is the granularity of the price dataset?'].apply(lambda x: 'day' if 'daily' in x.lower() else x)

# replace 'hourly' with 'hour'
consensus_df['What is the granularity of the price dataset?'] = consensus_df['What is the granularity of the price dataset?'].apply(lambda x: 'hour' if 'hourly' in x.lower() else x)

# show value counts for 'What is the granularity of the price dataset?'
print(consensus_df['What is the granularity of the price dataset?'].value_counts())
```
