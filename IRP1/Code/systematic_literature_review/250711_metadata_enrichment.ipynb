```python
import requests
import bibtexparser
from tqdm import tqdm

```

```python
def get_metadata_from_crossref(doi):
    url = f"https://api.crossref.org/works/{doi}"
    headers = {"User-Agent": "MySLRBot/1.0 (mailto:mathis.jander@utwente.nl)"}
    r = requests.get(url, headers=headers)
    if r.status_code != 200:
        return None
    msg = r.json()['message']
    return {
        'title': msg.get('title', [''])[0],
        'year': msg.get('issued', {}).get('date-parts', [[None]])[0][0],
        'journal': msg.get('container-title', [''])[0],
        'abstract': msg.get('abstract', ''),
        'keywords': ', '.join(msg.get('subject', [])),
        'authors': ['{} {}'.format(a.get('given', ''), a.get('family', '')) for a in msg.get('author', [])]
    }
```

```python


# Load .bib
with open("250711_final_corpus.bib") as bibtex_file:
    bib_database = bibtexparser.load(bibtex_file)

# Enrich each entry
for entry in tqdm(bib_database.entries):
    doi = entry.get('doi')
    if doi:
        meta = get_metadata_from_crossref(doi)
        if meta:
            # Convert authors from list to BibTeX-compatible string
            author_str = " and ".join(meta['authors'])  # e.g., "Alice Smith and Bob Jones"

            print("key words:", meta['keywords'])
            
            entry.update({
                'title': meta['title'],
                'year': str(meta['year']) if meta['year'] else '',
                'journal': meta['journal'],
                'abstract': meta['abstract'],
                'keywords': meta['keywords'],
                'author': author_str
            })

# Save enriched .bib
with open("enriched_final_corpus.bib", "w") as bibfile:
    bibtexparser.dump(bib_database, bibfile)
```
