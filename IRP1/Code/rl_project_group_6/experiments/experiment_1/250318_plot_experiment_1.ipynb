```python
import pandas as pd
from plot_functions import preprocess_df, plot_dual_line_per_episode, plot_per_episode_row
```

# Experiment 1

## 1) Experiments 1 - Results with SPARSE reward function

```python
# Load data
reward = "sparse"
df_logs = pd.read_csv(f"Results/training_logs_{reward}.csv")
df_logs = preprocess_df(df_logs)
# **1️⃣ Portfolio Value Plot**
plot_per_episode_row(df_logs, "new_portfolio_value", "Portfolio Value ($)", "Portfolio Value Across Episodes",
                     f"Results/{reward}_portfolio_value.png")

# **2️⃣ Portfolio Return Plot**
plot_per_episode_row(df_logs, "portfolio_return", "Portfolio Return (%)", "Portfolio Return Across Episodes",
                     f"Results/{reward}_portfolio_return_row.png")
# **3️⃣ Reward Plot**
plot_per_episode_row(df_logs, "reward", "Reward", "Reward Across Episodes",
                     f"Results/{reward}_reward_row.png")

# **4️⃣ Transaction Cost Plot**
plot_per_episode_row(df_logs, "transaction_cost", "Transaction Cost ($)", "Transaction Cost Across Episodes",
                     f"Results/{reward}_transaction_cost_row.png")
```

```python
# **1️⃣ Actions Plot**
plot_dual_line_per_episode(df_logs, "actions", "actions", 
                           "Allocation Weights Before Normalization", "Allocation Weights Before Normalization Over Time", 
                           "actions_plot.png", legend1="Action 1", legend2="Action 2")

# **2️⃣ Allocation Weights Plot**
plot_dual_line_per_episode(df_logs, "allocation_weights", "allocation_weights", 
                           "Allocation Weights", "Allocation Over Time", 
                           f"Results/{reward}_allocation_plot.png", 
                           legend1="Cash Weight", legend2="Stock Weight")

# **3️⃣ Stock Returns Plot**
plot_dual_line_per_episode(df_logs, "stock_returns", "stock_returns", 
                           "Stock Returns (%)", "Stock Returns Over Time", 
                           f"Results/{reward}_stock_returns_plot.png", 
                           legend1="Cash Return", legend2="Stock Return")

# **4️⃣ Old vs New Portfolio Value Plot (Using the Same Function)**
plot_dual_line_per_episode(df_logs, "old_portfolio_value", "new_portfolio_value", 
                           "Portfolio Value ($)", "Old vs New Portfolio Value Over Time", 
                           f"Results/{reward}_portfolio_value.png", 
                           legend1="Old Portfolio Value", legend2="New Portfolio Value")
```
