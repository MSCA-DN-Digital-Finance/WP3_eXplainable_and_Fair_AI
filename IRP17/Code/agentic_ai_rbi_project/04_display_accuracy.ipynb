```python
import json
from pathlib import Path
```

```python
RESULTS_PREFIX = "financebench_eval"
RESULTS_DIR = Path(".")
```

```python
def load_all_results():
    all_rows = []

    for f in RESULTS_DIR.glob(f"{RESULTS_PREFIX}*.json"):
        try:
            with f.open() as fh:
                data = json.load(fh)
                if isinstance(data, list):
                    all_rows.extend(data)
        except Exception as e:
            print(f"⚠️ Skipping {f.name}: {e}")

    return all_rows
```

```python
rows = load_all_results()

# Keep only rows that actually have a correctness label
judged = [
    r for r in rows
    if isinstance(r, dict) and "correct" in r
]

if not judged:
    print("No judged results found.")
 

total = len(judged)
correct = sum(1 for r in judged if r["correct"] == 1)
incorrect = total - correct

accuracy = correct / total

print("FINAL ACCURACY REPORT")
print("=" * 30)
print(f"Total evaluated: {total}")
print(f"Correct:         {correct}")
print(f"Incorrect:       {incorrect}")
print(f"Accuracy:        {accuracy:.4f} ({accuracy * 100:.2f}%)")
```
